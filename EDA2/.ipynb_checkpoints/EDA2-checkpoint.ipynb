```python
# Split Columns Based on Rule
onehot_cols = [col for col in cat_col
               if df[col].nunique() < 5 and col != "income"]

label_cols = [col for col in cat_col
              if df[col].nunique() >= 5 and col != "income"]

onehot_cols, label_cols
```

One-Hot Encoding

```python
df_encoded = df.copy()

df_encoded = pd.get_dummies(
    df_encoded,
    columns=onehot_cols,
    drop_first=True # Avoids dummy variable
)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in label_cols:
    df_encoded[col] = le.fit_transform(df_encoded[col])

# Remove leading/trailing spaces
df_encoded["income"] = df_encoded["income"].str.strip()

# encoding target Varible
df_encoded["income"] = df_encoded["income"].map({
    "<=50K": 0,
    ">50K": 1
})

# encode M/ F in 0,1
df_encoded["sex_ Male"] = df_encoded["sex_ Male"].astype(int)
```

Best practice

| Scenario                     | Recommended Encoding                    |
| ---------------------------- | --------------------------------------- |
| Nominal categorical variable | One-Hot Encoding                        |
| Ordinal categorical variable | Label Encoding                          |
| High-cardinality feature     | Label Encoding (or alternative methods) |
| Distance/linear model        | One-Hot Encoding                        |
| Tree-based model             | Label Encoding                          |


One-Hot Encoding — Advantages & Disadvantages

| Pros (Advantages)                                                                | Cons (Disadvantages)                                      |
| -------------------------------------------------------------------------------- | --------------------------------------------------------- |
| Eliminates ordinal bias — categories treated independently                       | Increases dimensionality (one column per category)        |
| Suitable for nominal categorical variables                                       | Can lead to curse of dimensionality                       |
| Works well with linear and distance-based models (SVM, kNN, Logistic Regression) | Higher memory and computational cost                      |
| Improves model interpretability (clear category representation)                  | Produces sparse matrices with many zeros                  |
| Prevents misleading numerical relationships between categories                   | Inefficient for high-cardinality features                 |
| Generally safe default encoding method                                           | May cause multicollinearity if all dummy columns retained |


Label Encoding — Advantages & Disadvantages

| Pros (Advantages)                                      | Cons (Disadvantages)                                          |
| ------------------------------------------------------ | ------------------------------------------------------------- |
| Simple and fast to implement                           | Introduces artificial ordinal relationship between categories |
| Keeps dimensionality low (single column)               | Misleads linear and distance-based models                     |
| Memory efficient                                       | Not suitable for nominal categorical variables                |
| Computationally inexpensive                            | Encoded values may distort feature importance                 |
| Appropriate for ordinal data (natural ordering exists) | Harder to interpret compared to one-hot representation        |
| Works well with tree-based models (Random Forest, GBM) | Mapping instability if categories change                      |


```python
df_fe = df_encoded.copy()
```

```python
# Identify individuals with any capital gain
df_fe["capital_gain_flag"] = (df_fe["capital_gain"] > 0).astype(int)

# Group individuals based on weekly working hour intensity
def working_hours_bucket(hours):
    if hours < 35:
        return "low"
    elif 35 <= hours <= 40:
        return "normal"
    else:
        return "high"

df_fe["working_hours_bucket"] = df_fe["hours_per_week"].apply(working_hours_bucket)

# Calculate overall investment outcome for individuals
df_fe["net_capital"] = df_fe["capital_gain"] - df_fe["capital_loss"]
```

The capital_gain_flag captures the presence of investment income, which is a strong indicator of higher earnings.
The working_hours_bucket and net_capital features summarize work intensity and overall investment outcome, reducing noise and providing more meaningful signals than raw values.

```python
df_fe[["capital_gain_flag", "working_hours_bucket", "net_capital"]]
```

Feature Engineering:

Capture hidden patterns

Improve prediction performance beyond raw features

```python
df_final = pd.get_dummies(
    df_fe,
    columns=["working_hours_bucket"],
    drop_first=True
)
df_final
```

```python
# Handle skewness
df_final["capital_gain_log"] = np.log1p(df_final["capital_gain"])
df_final["capital_loss_log"] = np.log1p(df_final["capital_loss"])
df_final["net_capital_log"] = np.log1p(df_final["net_capital"])
```

```python
print(df_final[['capital_gain','capital_loss','hours_per_week']].skew())
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
```

```python
num_col = df_final.select_dtypes(include=[np.number])
```

```python
num_col = num_col.dropna()
num_col
```

```python
# Fit Isolation Forest
iso = IsolationForest(
    n_estimators=100,
    contamination=0.02,   # assume 2% anomalies
    random_state=42
)
labels = iso.fit_predict(num_col)
```

```python
# Keep only normal points
df_no_outliers = df_final.loc[num_col.index][labels == 1]

print("Original shape:", df_final.shape)
print("After removing outliers:", df_no_outliers.shape)
```

The Isolation Forest algorithm was applied to numerical features to detect anomalous observations. The method isolates data points using random feature splits, identifying outliers as those requiring fewer partitions. Removing detected outliers reduced dataset size from X to Y records.

Outliers can negatively influence machine learning models by skewing parameter estimation, distorting distance metrics, increasing variance, and promoting overfitting. Eliminating extreme observations improves model robustness and generalization.

```python
df_pps = df_no_outliers.copy()

# Encode categoricals automatically
for col in df_pps.select_dtypes(include='object'):
    df_pps[col] = LabelEncoder().fit_transform(df_pps[col].astype(str))

target = 'income'

X = df_pps.drop(columns=[target])
y = df_pps[target]

model = RandomForestClassifier(n_estimators=50, random_state=42)

scores = {}

for col in X.columns:
    score = cross_val_score(
        model,
        X[[col]],
        y,
        cv=3,
        scoring='accuracy'
    ).mean()
    scores[col] = score

pps_like = pd.Series(scores).sort_values(ascending=False)

print(pps_like.head(10))
```

```python
# Correlation matrix (numeric features only)
corr_matrix = df_no_outliers.corr()
corr_matrix
```

Comparison of Feature Relationships:

The correlation matrix shows that education_num, capital_gain, net_capital, and hours_per_week have the strongest linear relationships with income, indicating their direct influence on earnings.

Engineered features such as capital_gain_flag, capital_gain_log, and net_capital_log exhibit very high correlation with their original variables, confirming that they are mathematically derived and not independent signals.

Several features (e.g., relationship, marital_status) show moderate negative correlation with income, highlighting social factors that affect earning patterns.

In comparison, the PPS-like analysis ranked some features higher even when their linear correlation was weak, demonstrating its ability to capture non-linear and asymmetric relationships that correlation cannot detect.

Overall, while the correlation matrix helps understand linear dependencies and redundancy, the PPS-like approach is more effective for feature selection and predictive modeling.
