#CHISQUARE Test

State the Hypotheses

`Null Hypothesis (H₀)`: No association between device type and customer satisfaction

`Alternative Hypothesis (H₁)`: There is an association between device type and customer satisfaction


```python
# input the observed frequencies from the contingency table and calculate
#expected frequencies assuming independence
import numpy as np
import pandas as pd
from scipy.stats import chi2

# Observed frequencies
observed = np.array([[50, 70], [80, 100],[60, 90],[30, 50],[20, 50]
])
```

```python
# Row and column totals
row_totals = observed.sum(axis=1)
row_totals
```

```python
col_totals = observed.sum(axis=0)
col_totals
```

```python
grand_total = observed.sum()
grand_total
```

```python
# Expected frequencies
expected = np.outer(row_totals, col_totals) / grand_total
expected
```

```python
# Chi-square statistic
chi_square_stat = ((observed - expected) ** 2 / expected).sum()
chi_square_stat = round(chi_square_stat, 2)
```

```python
chi_square_stat
```

 Determine the Critical Value

```python
# Degrees of freedom: (rows - 1) * (columns - 1)
df = (5 - 1) * (2 - 1)
alpha = 0.05
```

```python
# Critical value from chi-square distribution
critical_value = round(chi2.ppf(1 - alpha, df), 2)
critical_value
```

 Make a Decision

```python
# If the statistic is greater than the critical value, we reject the null hypothesis.
# If it is less, we accept the null hypothesis.

if chi_square_stat > critical_value:
    conclusion = "Reject the null hypothesis: There is a significant association."
else:
    conclusion = "accept the null hypothesis: No significant association."

conclusion
```

`Null Hypothesis (H₀)`:The weekly operating cost follows the theoretical model (W = $1,000 + $5X).  
`Alternative Hypothesis (H₁)`:The actual weekly operating cost is higher than the theoretical model.


Calculate the Test Statistic:

Use the following formula to calculate the test statistic (t):

where:
*   ˉxˉ = sample mean weekly cost (Rs. 3,050)
*   μ = theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)
*   σ = 5*25 units
*   n = sample size (25 restaurants)


```python
import math

# Given values
sample_mean = 3050
theoretical_mean = 1000 + 5 * 600
std_dev = 5 * 25
n = 25
```

```python
# Test statistic
t_stat = (sample_mean - theoretical_mean) / (std_dev / math.sqrt(n))
t_stat = round(t_stat, 2)
t_stat
```

Determine the Critical Value

```python
from scipy.stats import norm
alpha = 0.05
critical_value = round(norm.ppf(1 - alpha), 2)
critical_value
```

```python
#Make a Decision
if t_stat > critical_value:
    decision = "Reject the null hypothesis: There is strong evidence that actual costs are higher."
else:
    decision = "accept the null hypothesis: No strong evidence that actual costs are higher."

print(decision)
```
